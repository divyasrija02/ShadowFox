```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
df = pd.read_csv('titanic.csv')

print("Dataset Shape:", df.shape)

df.head()
```

```python
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df['Survived'].value_counts().plot(kind='bar')
plt.title("Survived vs Not Survived")
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Male vs Female Passengers")
plt.show()
```

# Titanic Dataset Analysis

## Objective
To analyze the Titanic dataset and understand passenger survival patterns.

## Findings
1. The dataset contains 891 passenger records.
2. Most passengers did not survive.
3. Male passengers were more than female passengers.
4. Some columns contain missing values.
5. Passenger class and gender affected survival rates.

## Conclusion
This project analyzed the Titanic dataset using Python, Pandas, and Matplotlib. Visualizations helped understand survival patterns among passengers.
